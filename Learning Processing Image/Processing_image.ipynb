```python
#importing the image module from matplotlib library
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
#loading the image from the path
img = mpimg.imread('cat_pic.jpg')
```

```python
type(img)
```

```python
print("Image shape:", img.shape)
```

```python
print(img)
```

```python
#displaying the image from numpy array using matplotlib
img_plot = plt.imshow(img)
plt.show()
```

Resizing the image using Pillow Library

```python
from PIL import Image
```

```python
img = Image.open('cat_pic.jpg')
img_resized = img.resize((200, 200))
img_resized.save('cat_pic_resized.jpg')
```

```python
#displaying the image from numpy array using matplotlib
img_res = mpimg.imread('cat_pic_resized.jpg')
img_res_plot = plt.imshow(img_res)
plt.show()
```

```python
print("Image shape:", img_res.shape)
```

**Converting RGB into Grayscale images using OpenCv**

```python
#import OpenCV library
import cv2
```

```python
img = cv2.imread('cat_pic.jpg')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
type(grayscale_img)
```

```python
grayscale_img.shape
```

```python
#displaying the image from numpy array using OpenCV
cv2.imshow('Grayscale Image', grayscale_img)
```
