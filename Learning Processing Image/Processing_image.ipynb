Resizing the image using Pillow Library

```python
from PIL import Image
```

```python
img = Image.open('cat_pic.jpg')
img_resized = img.resize((200, 200))
img_resized.save('cat_pic_resized.jpg')
```

```python
#displaying the image from numpy array using matplotlib
img_res = mpimg.imread('cat_pic_resized.jpg')
img_res_plot = plt.imshow(img_res)
plt.show()
```

```python
print("Image shape:", img_res.shape)
```

**Converting RGB into Grayscale images using OpenCv**

```python
#import OpenCV library
import cv2
```

```python
img = cv2.imread('cat_pic.jpg')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
type(grayscale_img)
```

```python
grayscale_img.shape
```

```python
# Safer than cv2.imshow inside Jupyter:
plt.imshow(grayscale_img, cmap='gray')
plt.axis('off')
plt.show()
```

```python
#displaying the image from numpy array using OpenCV
cv2.imshow('Grayscale Image', grayscale_img)
cv2.waitKey(0)          # wait for a keypress
cv2.destroyAllWindows() # close the window properly
```

```python
# saving the grayscale image
cv2.imwrite('cat_pic_grayscale.jpg', grayscale_img)
```

**Converting Grayscale into RGB:**

```python
#Read image as grayscale
gray = cv2.imread('cat_pic.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
rgb =cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
```

```python
rgb.shape
```

```python
plt.imshow(rgb)
```

Colorized the Grayscale image

```python
colored = cv2.applyColorMap(gray, cv2.COLORMAP_JET)
```

```python
colored_rgb = cv2.cvtColor(colored, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(colored_rgb)
```
